```python
import numpy as np
class DataAnalytics:

    title1 = "Numpy Analyzer"

    def __init__(self):
        self.arr1=None
        self.arr2=None

    @classmethod
    def welcome_msg(cls):
        print(f"Welcome to the {cls.title1}!")

    def _get_second_array(self):
        """Helper method to input a second 2D array of matching shape."""        
        total_elements = self.arr2.size
        user_input = input(f"\nEnter the same-size array elements ({total_elements} elements separated by space): ")

        data = list(map(int, user_input.split()))

        if len(data) != total_elements:
            print(f"Error: Please enter exactly {total_elements} elements!")
            return None
   
        second_arr = np.array(data).reshape(self.arr2.shape)
        return second_arr

    
    @staticmethod
    def exit_message():
        print("\nThank you for using the NumPy Analyzer! Goodbye!")
        
    def create_array(self):

            while True:
                print("\nSelect the type of array to create:")
                print("1. 1D Array")
                print("2. 2D Array")
                print("3. 3D Array")
                try:
                    choice = int(input("Enter your choice: "))

                    match choice:

                        case 1:
                            user1 = input("\nEnter elements for the array separated by space: ")
                            data = list(map(int, user1.split()))
                            self.arr1 = np.array(data)
                            print("\nArray created successfully:")
                            print(self.arr1)

                        case 2:
                            row1 = int(input("\nEnter the number of rows: "))
                            cols1 = int(input("Enter the number of columns: "))
                            user1 = input(f"Enter {row1 * cols1} elements for the array separated by space: ")
                            data = list(map(int, user1.split()))
                            self.arr2 = np.array(data).reshape(row1, cols1)
                            print("\nArray created successfully:")
                            print(self.arr2)
                            while True:
                                print("\nChoose an Operation:")
                                print("1. Indexing")
                                print("2. Slicing")
                                print("3. Go Back")
                                choice3=int(input("Enter your choice: "))
                                match choice3:
                                    case 1:
                                        row = int(input("\nEnter row index: "))
                                        col = int(input("Enter column index: "))
                                        print("Selected element:")
                                        print(self.arr2[row, col])
                                    case 2:
                                        row_range = input("\nEnter the row range (start:end): ")
                                        col_range = input("Enter the column range (start:end): ")

                                        row_start, row_end = map(int, row_range.split(":"))
                                        col_start, col_end = map(int, col_range.split(":"))

                                        print("\nSliced Array:")
                                        print(self.arr2[row_start:row_end, col_start:col_end])

                                    case 3:
                                        print("Going back...")
                                        break

                        case 3:
                            depth1 = int(input("\nEnter the number of depth:"))
                            row2 = int(input("Enter the number of rows:"))
                            cols2 = int(input("Enter the number of columns:"))
                            user2 = input( f"Enter {depth1 * row2 * cols2} elements for the array separated by space: ")
                            data = list(map(int, user2.split()))
                            self.arr1 = np.array(data).reshape(depth1, row2, cols2)
                            print("\nArray created successfully:")
                            print(self.arr1,"\n")

                            break

                        case _:
                            print("Invalid choice!")

                except ValueError:
                    print("Please enter only valid integer numbers!")
    def math_operation(self):
        if self.arr2 is None:
            print("\nPlease create an array first (Option 1).")
            return
        
        while True:
            print("\nChoose a mathematical operation:")
            print("1. Addition")
            print("2. Subtraction")
            print("3. Multiplication")
            print("4. Division")
            try:
                choice = int(input("Enter your choice: "))

                match choice:
                    case 1:  
                        second_arr = self._get_second_array()
                        if second_arr is not None:
                            print("\nOriginal Array:")
                            print(self.arr2)
                            print("\nSecond Array:")
                            print(second_arr)
                            print("\nResult of Addition:")
                            print(self.arr2 + second_arr)
                        else:
                            print("Please create a Second array!")
                    case 2:  
                        second_arr = self._get_second_array()
                        if second_arr is not None:
                            print("\nOriginal Array:")
                            print(self.arr2)
                            print("\nSecond Array:")
                            print(second_arr)
                            print("\nResult of Subtraction:")
                            print(self.arr2 - second_arr)
                        else:
                            print("Please create a Second array!")
                    case 3:  
                        second_arr = self._get_second_array()
                        if second_arr is not None:
                            print("\nOriginal Array:")
                            print(self.arr2)
                            print("\nSecond Array:")
                            print(second_arr)
                            print("\nResult of Multiplication:")
                            print(self.arr2 * second_arr)
                        else:
                            print("Please create a Second array!")

                    case 4:  
                        second_arr = self._get_second_array()

                        if second_arr is not None:
                            print("\nOriginal Array:")
                            print(self.arr2)

                            print("\nSecond Array:")
                            print(second_arr)

                            if np.any(second_arr == 0):
                                print("\nError: Division by zero is not allowed!")
                            else:
                                print("\nResult of Division:")
                                print(self.arr2 / second_arr, "\n")

                                break

                        else:
                            print("Please create a second array!")

                    case _:
                        print("Invalid choice! Please enter a valid option.")

            except ValueError:
                print("Please enter valid numeric input!")
    def combine_split_array(self):
        if self.arr2 is None:
            print("\nPlease create a 2D array first.")
            return
        
        while True:
            print("\nChoose an option:")
            print("1. Combine Arrays")
            print("2. Split Array")
            try:
                choice = int(input("Enter your choice: "))

                match choice:
                    case 1:
                        total = self.arr2.size
                        user_input = input(f"\nEnter the elements of another array to combine ({total} elements separated by space): ")
                        data = list(map(int, user_input.split()))

                        if len(data) != total:
                            print(f"Error: Exactly {total} elements required!")
                            continue

                        second_arr = np.array(data).reshape(self.arr2.shape)

                        print("\nOriginal Array:")
                        print(self.arr2)
                        print("\nSecond Array:")
                        print(second_arr)
                        print("\nCombined Array (Vertical Stack):")
                        print(np.vstack((self.arr2, second_arr)))
                    case 2:
                        print("\nOriginal Array:")
                        print(self.arr2)
                        splits = int(input("\nEnter number of splits: "))

                        result_split = np.vsplit(self.arr2, splits)
                        print("\nSplit Arrays:")
                        for part in result_split:
                            print(part)
                            print()

                        break

                    case _:
                        print("Invalid choice!")

            except ValueError:
                print("Error: Please enter valid integer numbers only!")

    def search_sort_filter(self):
        if self.arr2 is None:
            print("\nPlease create a 2D array first.")
            return
       
        while True:
            print("\nChoose an option:")
            print("1. Search a value")
            print("2. Sort the array")
            print("3. Filter values")
            try:
                choice = int(input("Enter your choice: "))

                match choice:
                    case 1:  
                        target = int(input("\nEnter the value to search: "))
                        rows, cols = np.where(self.arr2 == target)

                        if len(rows) > 0:
                            print(f"\nValue {target} found at:")
                            for i in range(len(rows)):
                                print(f"Row: {rows[i]}, Column: {cols[i]}")
                        else:
                            print(f"\nValue {target} not found in array.")

                    case 2: 
                        print("\nOriginal Array:")
                        print(self.arr2)

                        sorted_arr = np.sort(self.arr2)  
                        print("\nSorted Array:")
                        print(sorted_arr)
                        print("(Sorting applied row-wise.)")

                    case 3: 
                        val = int( input("\nEnter threshold value (Filter elements greater than): "))
                        filtered_arr = self.arr2[self.arr2 > val]

                        print(f"\nFiltered Values (Elements > {val}):")
                        print(filtered_arr,"\n")

                        break

                    case _:
                        print("Invalid choice!")

            except ValueError:
                print("Error: Please enter a valid integer number!")
            except Exception as e:
                print("Error:", e) 

    
    def aggregate_and_statistics(self):
        if self.arr2 is None:
            print("\nPlease create a 2D array first.")
            return
        print("\nChoose an aggregate/statistical operation:")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")

        while True:
            
            try:
                choice = int(input("Enter your choice: "))

                match choice:
                    case 1:
                        print("\nOriginal Array:")
                        print(self.arr2)
                        print("\nSum of Array:", np.sum(self.arr2))

                    case 2:
                        print("\nOriginal Array:")
                        print(self.arr2)
                        print("\nMean of Array:", np.mean(self.arr2))

                    case 3:
                        print("\nOriginal Array:")
                        print(self.arr2)
                        print("\nMedian of Array:", np.median(self.arr2))

                    case 4:
                        print("\nOriginal Array:")
                        print(self.arr2)
                        print("\nStandard Deviation of Array:", np.std(self.arr2))

                    case 5:
                        print("\nOriginal Array:")
                        print(self.arr2)
                        print("\nVariance of Array:", np.var(self.arr2),"\n")

                        break
                    

                    case _:
                        print("Invalid choice!")

            except ValueError:
                print("Please enter only valid integer numbers!")
            except Exception as e:
                print("Error:", e)

            
    def main_menu_func(self):
        """Main loop to control the user interface."""
        DataAnalytics.welcome_msg()
        print("====================================================")
        while True:
            print("Choose an option:")
            print("1. Create a Numpy Array")
            print("2. Perform Mathematical Operation")
            print("3. Combine or Split Arrays")
            print("4. Search, Sort or Filter Arrays")
            print("5. Compute Aggregates and Statistics")
            print("6. Exit")

            try:
                choice1 = int(input("Enter your choice: "))

                match choice1:
                    case 1:
                        self.create_array()
                    case 2:
                        self.math_operation()
                    case 3:
                        self.combine_split_array()
                    case 4:
                        self.search_sort_filter()
                    case 5:
                        self.aggregate_and_statistics()
                    case 6:
                        DataAnalytics.exit_message()

                        break
                    case _:
                        print("Invalid choice! Please select 1 to 6.")

            except ValueError:
                print("Error: Please enter a valid integer number!")
obj = DataAnalytics()
obj.main_menu_func()

        
```
